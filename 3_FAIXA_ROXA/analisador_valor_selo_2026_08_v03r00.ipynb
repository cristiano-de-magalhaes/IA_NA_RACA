````python
from ortools.linear_solver import pywraplp as pwl
````

```python
#CLASS VARIABLES for SOLUTION STATUS
pywraplp.Solver.OPTIMAL       = 0
pywraplp.Solver.FEASIBLE      = 1
pywraplp.Solver.INFEASIBLE    = 2
pywraplp.Solver.UNBOUNDED     = 3
pywraplp.Solver.ABNORMAL      = 4
pywraplp.Solver.MODEL_INVALID = 5
pywraplp.Solver.NOT_SOLVED    = 6
```

````python
# prova
print("pywraplp.Solver.OPTIMAL\t\t=", pwl.Solver.OPTIMAL)
print("pywraplp.Solver.FEASIBLE\t=", pwl.Solver.FEASIBLE)
print("pywraplp.Solver.INFEASIBLE\t=", pwl.Solver.INFEASIBLE)
print("pywraplp.Solver.UNBOUNDED\t=", pwl.Solver.UNBOUNDED)
print("pywraplp.Solver.ABNORMAL\t=", pwl.Solver.ABNORMAL)
print("pywraplp.Solver.MODEL_INVALID\t=", pwl.Solver.MODEL_INVALID)
print("pywraplp.Solver.NOT_SOLVED\t=", pwl.Solver.NOT_SOLVED)
````

````python
def solution_status(status):
    match status:
        case 0:
            print("OPTIMAL")
        case 1:
            print("FEASIBLE")
        case 2:
            print("INFEASIBLE")
        case 3:
            print("UNBOUNDED")
        case 4:
            print("ABNORMAL")
        case 5:
            print("MODEL_INVALID")
        case 6:
            print("NOT_SOLVED")
````

````python
#NumVar -> 100.00 para valores REAIS, ou seja, de ponto flutuante, ou "com vírgula"
#IntVar -> 100_000 para valores INTEIROS, ou seja, "sem vírgula"
````

## VM

````python
solver_vm = pwl.Solver(
    "Analisador de valor de selos VM", # Descrição
    pwl.Solver.GLOP_LINEAR_PROGRAMMING
)
````

````python
# NumVar(lb (lower boundary), ub (upper boundary), name)
selo_vm = solver_vm.NumVar(0.00, solver_vm.infinity(), "selo_vm")
````

````python
# RESTRIÇÕES VM

## faca descascar
# supondo que com a maior quantidade de selos o valor fica superior ao anunciado
solver_vm.Add(249.99 <= 65 * selo_vm +  9.99, "faca_dascascar_1")
# supondo que com a menor quantidade de selos o valor fica inferior ao anunciado
solver_vm.Add(249.99 >= 25 * selo_vm + 59.99, "faca_dascascar_2")

# faca utilitária
solver_vm.Add(263.99 <= 65 * selo_vm + 19.99, "faca_utilitaria_1")
solver_vm.Add(263.99 >= 25 * selo_vm + 69.99, "faca_utilitaria_2")

# faca pão
solver_vm.Add(372.99 <= 75 * selo_vm + 19.99, "faca_pao_1")
solver_vm.Add(372.99 >= 35 * selo_vm + 69.99, "faca_pao_2")

# faca chefe
solver_vm.Add(395.99 <= 75 * selo_vm + 24.99, "faca_chefe_1")
solver_vm.Add(395.99 >= 35 * selo_vm + 79.99, "faca_chefe_2")

# faca santoku
solver_vm.Add(395.99 <= 75 * selo_vm + 24.99, "faca_santoku_1")
solver_vm.Add(395.99 >= 35 * selo_vm + 79.99, "faca_santoku_2")

# tesoura
solver_vm.Add(432.99 <= 80 * selo_vm + 24.99, "tesoura_1")
solver_vm.Add(432.99 >= 40 * selo_vm + 79.99, "tesoura_2")

# tábua
solver_vm.Add(499.99 <= 80 * selo_vm + 39.99, "tabua_1")
solver_vm.Add(499.99 >= 35 * selo_vm + 89.99, "tabua_2")

# # cepo *
# solver_vm.Add(665.99 <= 80 * selo_vm + 49.99, "cepo_1")
# solver_vm.Add(665.99 >= 35 * selo_vm + 99.99, "cepo_2")
````

````python
solver_vm.Minimize(selo_vm)
````

````python
status_vm = solver_vm.Solve()
solution_status(status_vm)
````

````python
print("Selo: R$", f"{selo_vm.solution_value():.2f}")
print("Selo: R$", f"{solver_vm.Objective().Value():.2f}")
````

````python
valor_selo_vm = selo_vm.solution_value()
````
